```python
import pandas as pd
from sklearn.model_selection import train_test_split
import torch
from torch.utils.data import DataLoader,Dataset
import torch.nn as nn
import torch.optim as optim
import os
from google.colab import drive #type:ignore
drive.mount(r'/content/drive/')
df = pd.read_csv(r"/content/drive/MyDrive/DL_CSV/fmnist_small.csv")
x=df.iloc[:,1:].values
y=df.iloc[:,0].values

x=x/255.0

print(x)
print(y)

#creating custom class dataset
class CustomDataset(Dataset):

    def __init__(self,features,labels):
        self.features = torch.tensor(features, dtype=torch.float32) #nn.Linear layers have internal weights that are 32-bit floats. Passing a default NumPy float64 array directly into the model causes a crash
        self.labels = torch.tensor(labels, dtype=torch.long) #nn.CrossEntropyLoss() specifically demands that target labels be integers of type int64 (known as torch.long in PyTorch 

    def __len__(self):
        return len(self.features)     

    def __getitem__(self, index):
        return self.features[index], self.labels[index]
    
#Performing DataSet Split
X_train, X_test, y_train, y_test =train_test_split(x,y,test_size=0.2,random_state=42,shuffle=True)

#Creating CustomDataset Object for train and test
train_dataset=CustomDataset(X_train,y_train)
test_dataset=CustomDataset(X_test,y_test)

#Creating DataLoader object of the class
train_loader=DataLoader(train_dataset,batch_size=32,shuffle=True,num_workers=2)
test_loader=DataLoader(test_dataset,batch_size=32,shuffle=False)

#Creating our Artificial Neural Network
class ArtificialNN(nn.Module):

    def __init__(self,num_features):
        super().__init__()
        self.network=nn.Sequential(
            nn.Linear(num_features,128),
            nn.ReLU(),
            nn.Linear(128,64),
            nn.ReLU(),
            nn.Linear(64,10)
        )

    def forward(self,features):
        return self.network(features)
        
#Defining Epochs and Learning Rate
epochs=100
learning_rate=0.1

model=ArtificialNN(X_train.shape[1])
loss_function=nn.CrossEntropyLoss()
optimizer=optim.SGD(model.parameters(),lr=learning_rate)

#Performing Training
for i in range(epochs):

    total_loss=0

    for featues,label in train_loader:
        y_pred=model(featues)

        #Making gradient zero
        optimizer.zero_grad()

        #Calculating Loss
        loss=loss_function(y_pred,label)

        #Backward
        loss.backward()

        #Updating Paramters
        optimizer.step()

        total_loss=total_loss+loss.item()
    
    print(f"For epoch :{i+1}, Loss={total_loss/len(train_loader)}")

#Evaluating the model
model.eval() #Putting the model in evaluating modde
total=0
correct=0

with torch.no_grad():
    for features,labels in test_loader:
        y_pred=model(features)

        _,predicted=torch.max(y_pred,1)
        total=total+labels.shape[0]

        correct=correct+(predicted==labels).sum().item()

print(f"Accuracy of Model = {(correct/total)*100:.2f}%")
```
